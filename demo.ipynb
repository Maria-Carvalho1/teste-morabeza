```python
import pandas as pd

vendas = pd.read_csv('data/tabelas/vendas.csv', parse_dates=["data"])
clientes = pd.read_csv('data/tabelas/clientes.csv')
produtos = pd.read_csv('data/tabelas/produtos.csv')
vendas.head(2)
```

```python
df1 = vendas.merge(clientes[["cliente_id", "tipo", "ilha"]], 
                on="cliente_id", 
                how="left", 
                validate="many_to_one").merge(produtos[["produto_id", "nome", "categoria"]],
                                              on="produto_id", 
                                              how="left")
```
